```python
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Load pre-trained BART model and tokenizer for summarization
model_name = "facebook/bart-large-cnn"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
# Define input article text
article = """
Natural Language Processing (NLP) is a subfield of artificial intelligence (AI)
that focuses on the interaction between computers and human language.
Modern NLP leverages deep learning models, such as Transformers, to perform complex
tasks like text generation, translation, sentiment analysis, and summarization.
With recent advances in Large Language Models (LLMs), businesses are increasingly
integrating RAG architectures and fine-tuned models to automate workflows.
"""

# Tokenize input text
inputs = tokenizer(article, max_length=1024, return_tensors="pt", truncation=True)

# Generate summary ids using model
summary_ids = model.generate(
    inputs["input_ids"],
    num_beams=4,
    min_length=20,
    max_length=60,
    early_stopping=True
)

# Decode generated token ids to text
summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

print("Original Article:\n", article)
print("\nGenerated Summary:\n", summary)
```
